```python
def g(x):
    return x**3 + 2*x

# 1. Numerical derivative at x = 2 (use h = 1e-6)
h = 1e-6
slope_at_2=round((g(2+h)-g(2))/h,4)
print(slope_at_2)

# 2. Symbolic derivative of x**3 + 2*x
import sympy as sp
x=sp.symbols('x')
expr= x**3 + 2*x
deriv=sp.diff(expr,x)
print(deriv)

# 3. Evaluate the symbolic derivative at x = 2 and compare
sym=float(deriv.subs(x,2))
print(sym)
if sym==slope_at_2:
  print("True")
else:
  print("False")
```
